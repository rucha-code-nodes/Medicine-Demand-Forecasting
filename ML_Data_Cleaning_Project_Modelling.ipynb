```python
#DATA CLEANING

import pandas as pd
import numpy as np

# Ensure 'df' is defined in the current runtime session by running your first cell before executing this one.
print(f"Original shape before cleaning: {df.shape}")

# 1. Remove exact duplicate rows
df = df.drop_duplicates()
print(f"Shape after removing duplicates: {df.shape}")

# 2. Clean 'NULL' text strings to actual NaN values across the dataframe
df = df.replace(['NULL', 'null', 'None', 'none', 'NaN', 'nan'], np.nan)

# 3. Standardize the Date column (Replace 'Date' with your actual date column name if different)
# We use errors='coerce' so bad date formats automatically turn into NaT (Not a Time)
date_col = 'Date' # Adjust if your column name is slightly different
if date_col in df.columns:
    df[date_col] = pd.to_datetime(df[date_col], errors='coerce')

# 4. Drop rows where critical columns (Date, Clinic, Medicine) are missing
critical_cols = [date_col, 'Clinic', 'Medicine']
# Check which of these columns actually exist in your dataframe
existing_critical_cols = [col for col in critical_cols if col in df.columns]

if existing_critical_cols:
    df = df.dropna(subset=existing_critical_cols)
    print(f"Shape after dropping rows missing critical data ({existing_critical_cols}): {df.shape}")

# 5. Clean text columns (strip extra whitespace from Clinic, Medicine, and Channel Name)
# Adjusted 'ChannelName' to 'Channel Name' to match the loaded columns in your dataset
for col in ['Clinic', 'Medicine', 'Channel Name']:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip()

        # Step: Fill secondary/text columns with 'Unknown' instead of dropping them
text_columns_to_fill = ['Vitals', 'ChiefComplaint', 'Diagnosis', 'PatientState', 'Advice', 'ChannelName']

for col in text_columns_to_fill:
    if col in df.columns:
        df[col] = df[col].fillna('Unknown')

print("Optional text columns filled successfully!")

# Check if any missing values remain in important columns
print("\nRemaining missing values check:")
print(df.isnull().sum()[df.isnull().sum() > 0])
print(f"\nFinal Ready Dataset Shape: {df.shape}")

print("\nData cleaning complete!")
print(f"Final clean dataset shape: {df.shape}")
```

```python
#Aggregating Data into Monthly Demand
import pandas as pd

# Step 1: Ensure the Date column is datetime type
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

# Step 2: Extract Year and Month (e.g., '2023-05')
df['year_month'] = df['Date'].dt.to_period('M')

# Step 3: Group by Clinic, Medicine, and Month, then count the total prescriptions/usage
monthly_demand = df.groupby(
    ['Clinic', 'Medicine', 'year_month']
).size().reset_index(name='quantity')

# Step 4: Convert the period back to a standard timestamp for modeling
monthly_demand['year_month'] = monthly_demand['year_month'].dt.to_timestamp()

# Preview the results
print(f"Total rows in monthly demand table: {monthly_demand.shape[0]}")
monthly_demand.head(10)
```

```python
#Step 5: Feature Engineering (Creating Lag Features)
# Step 1: Sort the monthly demand data chronologically by Clinic, Medicine, and Month
monthly_demand = monthly_demand.sort_values(by=['Clinic', 'Medicine', 'year_month'])

# Step 2: Create Lag 1 (Demand from 1 month ago) for each clinic and medicine
monthly_demand['lag_1'] = monthly_demand.groupby(['Clinic', 'Medicine'])['quantity'].shift(1)

# Step 3: Create Lag 2 (Demand from 2 months ago)
monthly_demand['lag_2'] = monthly_demand.groupby(['Clinic', 'Medicine'])['quantity'].shift(2)

# Step 4: Extract the Month (1 to 12) as a seasonal feature
monthly_demand['month'] = monthly_demand['year_month'].dt.month

# Step 5: Drop rows with NaN values created by shifting (since the first two months won't have past history)
monthly_demand = monthly_demand.dropna(subset=['lag_1', 'lag_2'])

# Preview the feature-engineered dataset
print(f"Total rows after creating lag and seasonal features: {monthly_demand.shape[0]}")
display(monthly_demand.head(10))
```

```python
#Step 6: Model Training & Evaluation.

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

print("Preparing features and target...")

# 1. Define Features (Inputs) and Target (What we want to predict)
features = ['lag_1', 'lag_2', 'month']
target = 'quantity'

X = monthly_demand[features]
y = monthly_demand[target]

# 2. Time-based split (Using the last 20% of data as the test set to mimic future prediction)
split_index = int(len(monthly_demand) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]
y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

print(f"Training set shape: {X_train.shape}")
print(f"Testing set shape: {X_test.shape}")

# 3. Initialize and Train the Random Forest Model
print("\nTraining the Random Forest model (this might take a moment)...")
model = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
model.fit(X_train, y_train)
print("Model training complete!")

# 4. Make Predictions on the Test Set
print("\nEvaluating model performance...")
predictions = model.predict(X_test)

# 5. Calculate Errors
mae = mean_absolute_error(y_test, predictions)
rmse = np.sqrt(mean_squared_error(y_test, predictions))

print(f"=== MODEL EVALUATION RESULTS ===")
print(f"Mean Absolute Error (MAE): {mae:.2f} units")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f} units")
```

```python
import pickle

# Save the trained model to Google Drive/Colab local storage
model_filename = 'medicine_demand_model.pkl'
with open(model_filename, 'wb') as file:
    pickle.dump(model, file)

print(f"Model successfully saved as '{model_filename}'!")
```

```python
def predict_medicine_demand(lag_1, lag_2, month):
    """
    Predicts next month's medicine demand given past demand and the target month.
    """
    # Create the input format expected by the model
    user_input = [[lag_1, lag_2, month]]

    # Make the prediction
    prediction = model.predict(user_input)[0]

    # Round to a sensible number (since medicine quantities are whole units)
    return round(prediction)

# --- EXAMPLE TEST ---
# Suppose a clinic used 10 units last month (lag_1), 8 units two months ago (lag_2),
# and we want to predict demand for July (month = 7)
sample_prediction = predict_medicine_demand(lag_1=10, lag_2=8, month=7)

print(f"Predicted Demand for next month: {sample_prediction} units")
```

```python
%%writefile app.py
import streamlit as st
import pandas as pd
import pickle

# 1. Load the trained model and lookup data
@st.cache_resource
def load_resources():
    with open('medicine_demand_model.pkl', 'rb') as f:
        model = pickle.load(f)

    df_lookup = pd.read_pickle('monthly_demand_lookup.pkl')
    return model, df_lookup

model, df_lookup = load_resources()

st.title("💊 Medicine Demand Forecasting System")
st.write("Predict monthly medicine requirements for clinics based on historical consumption.")

# 2. User Selection Dropdowns
clinics = df_lookup['Clinic'].unique().tolist()
selected_clinic = st.selectbox("Select Clinic", clinics)

# Filter medicines available for that clinic
medicines = df_lookup[df_lookup['Clinic'] == selected_clinic]['Medicine'].unique().tolist()
selected_medicine = st.selectbox("Select Medicine / Bundle", medicines)

# Select target month (1 to 12)
target_month = st.slider("Select Target Month (1=Jan, 12=Dec)", min_value=1, max_value=12, value=6)

# 3. Automatically fetch the latest available lag data for this clinic & medicine
record = df_lookup[(df_lookup['Clinic'] == selected_clinic) & (df_lookup['Medicine'] == selected_medicine)]

if not record.empty:
    # Get the most recent lag values from the dataset
    latest_row = record.iloc[-1]
    lag_1 = latest_row['lag_1']
    lag_2 = latest_row['lag_2']

    st.write(f"---")
    st.write(f"**Historical Context Found:**")
    st.write(f"- Demand last month (`lag_1`): {lag_1} units")
    st.write(f"- Demand two months ago (`lag_2`): {lag_2} units")

    # 4. Predict Button
    if st.button("Generate Forecast"):
        user_input = pd.DataFrame([[lag_1, lag_2, target_month]], columns=['lag_1', 'lag_2', 'month'])
        prediction = model.predict(user_input)[0]
        final_prediction = max(0, round(prediction))

        st.success(f"📦 Predicted Demand for Next Month: **{final_prediction} units**")
else:
    st.warning("No historical data found for this specific combination.")
```
